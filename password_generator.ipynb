```python
import secrets
import string
import hashlib


###Password Generation and Hashing Program


##Password Generation

print("Password Generator")

print("\nTypically, the following information would be private. However, to prove this code works, the following is printed:")

print("\nGenerating Password...\n")

password = []

size = None

size_integer = secrets.randbelow(20) #generates a random integer from 0 up to, but not including 20
    
if 0 <= size_integer <= 8:
    size = secrets.choice(range(8, 16)) #if the random integer called "size_integer" is between 0 to 8, set password size somewhere bewteen 8 to 16 characters
        
if 9 <= size_integer <= 12:
    size = secrets.choice(range(16, 25))
        
if 13 <= size_integer <= 16:
    size = secrets.choice(range(25, 40))

if 17 <= size_integer <= 19:
    size = secrets.choice(range(40, 65))


#loop through the random integers generated below until all password characters have been selected based on its size
for i in range(size):
    password_integers = secrets.randbelow(20) #generates a random integer from 0 up to, but not including 20
    
    if 0 <= password_integers <= 8:
        lowercase = secrets.choice(string.ascii_lowercase)
        password.append(lowercase) #if the random integer called "passwords_integers" is between 0 to 8, make one of the password's character's lowercase
        
    if 9 <= password_integers <= 12:
        uppercase = secrets.choice(string.ascii_uppercase)
        password.append(uppercase)
        
    if 13 <= password_integers <= 16:
        digits = secrets.choice(string.digits)
        password.append(digits)

    if 17 <= password_integers <= 19:
        punctuation = secrets.choice(string.punctuation)
        password.append(punctuation)

secrets.SystemRandom().shuffle(password) #shuffle the generated password

password = "".join(password) #combine characters into a single "word" instead of a having series of characters

print("Password: ")
print(password)


##Salting tool

print("\nGenerating and appending salt value to your password...")
salt_value = secrets.token_hex(16) #generates 32 hexadecimal characters, representing 16 random bytes

salted_password = password + salt_value

print("\nSalted Password:")
print(salted_password)

#Hashing tool
print("\nHashing password...")

salted_password_bytes = salted_password.encode() #convert salted password into bytes

hashed_password = hashlib.pbkdf2_hmac('sha256', salted_password_bytes, b'', 600000)
#pdkdf2_hmac() takes a password and produces a password hash using Password-Based Key Derivation Function 2 (PBKDF2)
#sha256 is the PBKDF2 hashing algorithm used
#b'' means our string/bytes argument is empty for this function

hashed_password = hashed_password.hex() #converts byte formatting to hexidecimal

print("\nHashed Password")
print(hashed_password)
























```
